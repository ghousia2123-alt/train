```python
distance = float(input("Enter distance in km: "))
age = int(input("Enter passenger age: "))

if distance <= 100:
    fare = distance * 2
elif distance <= 300:
    fare = distance * 1.5
else:
    fare = distance * 1

if age < 5:
    fare = 0
elif age >= 60:
    fare = fare * 0.5

print("Train Ticket Fare =", fare)
```
